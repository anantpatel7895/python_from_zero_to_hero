# Level 6: Stacking decorators

You can apply several decorators to one function. They are applied **bottom-up** (the closest to `def` first), but when the function is called they run **top-down**, like layers of an onion.

```python
@A
@B
def f(): ...
# same as: f = A(B(f))
```

````python
import functools
import time
import random
````

````python
def bold(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        return "<b>" + func(*args, **kwargs) + "</b>"
    return wrapper

def italic(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        return "<i>" + func(*args, **kwargs) + "</i>"
    return wrapper

@bold
@italic
def text():
    return "hello"

@italic
@bold
def text2():
    return "hello"

print(text())    # bold(italic(text))
print(text2())   # italic(bold(text2))
````

Watch the order with print statements:

````python
def layer(name):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            print(f"enter {name}")
            result = func(*args, **kwargs)
            print(f"exit  {name}")
            return result
        return wrapper
    return decorator

@layer("outer")
@layer("middle")
@layer("inner")
def core():
    print("  core runs")

core()
````

> **Why it matters:** order changes behaviour. For example, `@timer` above `@retry` measures total time including retries; `@timer` below `@retry` measures each attempt.
