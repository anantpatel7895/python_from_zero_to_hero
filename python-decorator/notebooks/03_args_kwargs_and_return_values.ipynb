# Level 2: General-purpose wrappers

## 2.1 The problem: arguments
Our `wrapper()` takes no arguments, so it breaks on a function that needs some.

````python
# my_decorator from the previous notebook
def my_decorator(func):
    def wrapper():
        print("Before the function runs")
        func()
        print("After the function runs")
    return wrapper
````

````python
@my_decorator
def add(a, b):
    return a + b

try:
    add(2, 3)
except TypeError as e:
    print("Error:", e)
````

## 2.2 The fix: `*args` and `**kwargs`, and return the result

- `*args` collects any positional arguments into a tuple.
- `**kwargs` collects any keyword arguments into a dict.
- The wrapper must `return` what the original function returned, or the result is lost.

````python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__} with args={args}, kwargs={kwargs}")
        result = func(*args, **kwargs)
        print(f"{func.__name__} returned {result}")
        return result
    return wrapper

@my_decorator
def add(a, b):
    return a + b

@my_decorator
def power(base, exp=2):
    return base ** exp

print(add(2, 3))
print(power(3, exp=3))
````

## 2.3 The decorator template

Memorise this. Almost every decorator starts from it:

```python
import functools

def decorator(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        # do something BEFORE
        result = func(*args, **kwargs)
        # do something AFTER
        return result
    return wrapper
```
