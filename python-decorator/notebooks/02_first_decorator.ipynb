# Level 1: Your first decorator

**A decorator is a function that takes a function and returns a new function that usually adds behaviour around the original.**

```
original function  ──►  decorator  ──►  new "wrapped" function
```

## 1.1 Decorating by hand

````python
def my_decorator(func):
    print("Inside my_decorator")
    def wrapper():
        print("Before the function runs")
        func()
        print("After the function runs")
    return wrapper

def say_hello():
    print("Hello!")

say_hello = my_decorator(say_hello)   # replace the function with the wrapped one
say_hello()
````

## 1.2 The `@` syntax

`@my_decorator` placed above a function is shorthand for `say_hello = my_decorator(say_hello)`. It does exactly the same thing.

````python
@my_decorator
def say_bye():
    print("Bye!")

say_bye()
````

> **Key point:** the decorator runs **once, when the function is defined**. The `wrapper` runs **every time you call** the function. Check it below:

````python
say_bye()
````

> clearly can see that the decorator runs only once, when the function is defined, and the wrapper runs every time you call the function. for example: when we second time call the function, **Inside my_decorator ** is not printed, but **Before the function runs** and **Hello!** are printed.


````python
def announce(func):
    print(f"[decorating {func.__name__}]")   # runs at definition time
    def wrapper():
        print("[calling]")                   # runs at call time
        return func()
    return wrapper

@announce
def task():
    print("doing task")

print("--- now calling ---")
task()
task()
````
