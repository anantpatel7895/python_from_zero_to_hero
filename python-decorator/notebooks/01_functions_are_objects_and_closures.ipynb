# Level 0: Foundations

A decorator is built from three ideas. Get these clear first and decorators become easy.

## 0.1 Functions are objects

In Python a function is a value, like an `int` or a `list`. You can store it in a variable, put it in a list, or pass it to another function.

````python
def greet(name):
    return f"Hello, {name}!"

say_hi = greet          # no parentheses: we are NOT calling it, we are referring to it
print(say_hi("Anant"))
print(type(greet))
print(greet.__name__)
````

## 0.2 Passing a function to another function

A function that takes (or returns) another function is called a **higher-order function**.

````python
def shout(text):
    return text.upper()

def whisper(text):
    return text.lower()

def speak(style_func, message):
    return style_func(message)

print(speak(shout, "Hello"))
print(speak(whisper, "Hello"))
````

## 0.3 Functions defined inside functions

You can define a function inside another function. The inner one exists only while the outer one runs, unless you return it.

````python
def outer():
    def inner():
        return "I am inner"
    return inner          # return the function itself, not inner()

f = outer()
print(f)
print(f())
````

## 0.4 Closures: inner functions remember outer variables (IMPORTANT)

The inner function **remembers** the variables of the outer function, even after the outer function has finished. This is called a **closure**, and it is the engine that makes decorators work.

````python
def make_multiplier(n):
    def multiply(x):
        return x * n      # n comes from the outer function
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)

print(double(5))   # 10
print(triple(5))   # 15
````

````python
def greet(name):
    return f"Hello, {name}!"

say_hi = greet           # a second label for the same function

print(say_hi.__name__)   # 'greet', not 'say_hi'
print(say_hi is greet)   # True: one object, two names

del greet                # remove the label 'greet'
print(say_hi("Anant"))   # still works: the object is still alive
print(say_hi.__name__)   # still 'greet'
````
