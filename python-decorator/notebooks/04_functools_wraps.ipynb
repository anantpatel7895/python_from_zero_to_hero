# Level 3: `functools.wraps`

## 3.1 The problem: the function loses its identity

```python
# my_decorator from the previous notebook (no functools.wraps yet)
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__} with args={args}, kwargs={kwargs}")
        result = func(*args, **kwargs)
        print(f"{func.__name__} returned {result}")
        return result
    return wrapper
```

```python
@my_decorator
def multiply(a, b):
    """Multiply two numbers."""
    return a * b

print(multiply.__name__)   # expected 'multiply'...
print(multiply.__doc__)    # expected the docstring...
```

The name and docstring now belong to `wrapper`. This breaks `help()`, debugging, logging and tools that inspect functions.

## 3.2 The fix
`functools.wraps(func)` copies the name, docstring, module and other metadata from `func` onto `wrapper`. It also sets `__wrapped__`, which points to the original function.

```python
import functools

def my_decorator(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper

@my_decorator
def multiply(a, b):
    """Multiply two numbers."""
    return a * b

print(multiply.__name__)
print(multiply.__doc__)
print(multiply.__wrapped__)      # the original, undecorated function
```

> **Rule:** always use `@functools.wraps(func)` on your wrapper.
