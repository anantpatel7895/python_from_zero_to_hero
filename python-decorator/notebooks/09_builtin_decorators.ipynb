# Level 8: Built-in decorators you will use every day

## 8.1 `@property`, `@staticmethod`, `@classmethod`

```python
import functools
import time
import random
```

```python
class Temperature:
    unit_symbol = "C"

    def __init__(self, celsius):
        self._celsius = celsius

    @property
    def celsius(self):
        """Read like an attribute: t.celsius"""
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Below absolute zero!")
        self._celsius = value

    @property
    def fahrenheit(self):
        return self._celsius * 9 / 5 + 32

    @staticmethod
    def is_freezing(celsius):
        """Does not need self or cls: just a function that lives in the class."""
        return celsius <= 0

    @classmethod
    def from_fahrenheit(cls, f):
        """Receives the class: often used as an alternative constructor."""
        return cls((f - 32) * 5 / 9)

t = Temperature(25)
print(t.celsius, t.fahrenheit)
t.celsius = 30
print(t.celsius)
print(Temperature.is_freezing(-5))
t2 = Temperature.from_fahrenheit(212)
print(round(t2.celsius, 2))

try:
    t.celsius = -300
except ValueError as e:
    print("Error:", e)
```

## 8.2 `@functools.lru_cache`: memoisation for free

It caches results, so repeated calls with the same arguments are instant. Classic example: Fibonacci.

```python
def fib_slow(n):
    return n if n < 2 else fib_slow(n - 1) + fib_slow(n - 2)

@functools.lru_cache(maxsize=None)
def fib_fast(n):
    return n if n < 2 else fib_fast(n - 1) + fib_fast(n - 2)

start = time.perf_counter(); fib_slow(30); print(f"slow: {time.perf_counter() - start:.3f}s")
start = time.perf_counter(); fib_fast(30); print(f"fast: {time.perf_counter() - start:.6f}s")
print(fib_fast(100))
print(fib_fast.cache_info())
```

## 8.3 Build your own cache to see how it works

```python
def memoize(func):
    cache = {}
    @functools.wraps(func)
    def wrapper(*args):
        if args not in cache:
            cache[args] = func(*args)
        return cache[args]
    wrapper.cache = cache
    return wrapper

@memoize
def fib(n):
    return n if n < 2 else fib(n - 1) + fib(n - 2)

print(fib(50))
print(len(fib.cache), "values cached")
```
