```python
class Dog:
    def __init__(self, name):
        self.name = name

    def bark(self):
        print(f"{self.name} says woof!")

d1 = Dog("Rex")   # d1 is a reference to a new Dog object in memory
d2 = Dog("Fido")  # d2 is a reference to a different Dog object
```

```python
print(type(d1))
```

```python
print(d1.__name__)  # Output: Rex
```
